Fila en Caja (FIFO): Atender clientes en estricto orden de llegada usando collections.deque con popleft(). Validar qué sucede cuando se intenta cobrar y no hay nadie en la fila.

```python
from collections import deque

# 1. Crear la fila (deque) vacía
fila_caja = deque()

# 2. Llegada de clientes (se agregan por la derecha)
fila_caja.append("Carlos")
fila_caja.append("Ana")
fila_caja.append("Luis")

print(f"Fila inicial: {list(fila_caja)}")

# 3. Atender clientes en orden FIFO (se sacan por la izquierda con popleft)
print(f"Atendiendo a: {fila_caja.popleft()}")
print(f"Atendiendo a: {fila_caja.popleft()}")
print(f"Atendiendo a: {fila_caja.popleft()}")

# En este punto la fila ya está vacía
print(f"Fila actual: {list(fila_caja)}")

# 4. Validar qué sucede al intentar cobrar cuando no hay nadie en la fila
if len(fila_caja) > 0:
    cliente = fila_caja.popleft()
    print(f"Atendiendo a: {cliente}")
else:
    print("Error: No se puede cobrar porque no hay nadie en la fila.")
```

Bolsa de Productos (LIFO): Cada cliente tiene una bolsa (pila). Al momento de pagar, los productos se ingresan en un orden, pero el cajero debe desempacar o registrar primero el último producto guardado (pop()).

```python
print("\n--- Simulando Bolsa de Productos (LIFO) ---")

# 1. El cliente ingresa sus productos a la bolsa (pila)
# Se usa una lista estándar de Python donde el final representa "arriba" de la bolsa
bolsa_cliente = []

bolsa_cliente.append("Leche")
bolsa_cliente.append("Pan")
bolsa_cliente.append("Manzana")  # Este fue el último producto guardado

print(f"Productos guardados en la bolsa (orden de ingreso): {bolsa_cliente}")

# 2. El cajero registra los productos sacándolos desde arriba de la bolsa (LIFO)
# pop() elimina y devuelve el último elemento de la lista
print(f"Cajero registra: {bolsa_cliente.pop()}")
print(f"Cajero registra: {bolsa_cliente.pop()}")
print(f"Cajero registra: {bolsa_cliente.pop()}")

# 3. Validar qué sucede si se intenta sacar un producto de la bolsa vacía
if len(bolsa_cliente) > 0:
    producto = bolsa_cliente.pop()
    print(f"Cajero registra: {producto}")
else:
    print("Error: No se pueden registrar más productos, la bolsa está vacía.")
```

Módulos de Cupones y Membresías (Sets): Manejar un conjunto de códigos de membresía vigentes. Registrar el cupón usado para evitar que se repita en la misma sesión

```python
print("\n--- Simulando Cupones y Membresías (Sets) ---")

# 1. Conjunto de membresías vigentes en la tienda
membresias_vigentes = {"VIP123", "PREMIUM456", "BASIC789"}

# 2. Conjunto vacío para registrar los cupones que ya se usaron en la sesión
cupones_usados = set()

# Caso A: Validar una membresía
codigo_cliente = "VIP123"
if codigo_cliente in membresias_vigentes:
    print(f"Membresía {codigo_cliente} válida. Aplicando descuento base.")
else:
    print(f"Membresía {codigo_cliente} no es válida.")

# Caso B: Registrar y validar uso de un cupón (Primera vez)
cupón_ingresado = "DESC20"

if cupón_ingresado in cupones_usados:
    print(f"Error: El cupón {cupón_ingresado} ya fue utilizado en esta sesión.")
else:
    cupones_usados.add(cupón_ingresado)
    print(f"Cupón {cupón_ingresado} aplicado con éxito.")

# Caso C: Intentar registrar el mismo cupón otra vez
cupón_repetido = "DESC20"

if cupón_repetido in cupones_usados:
    print(f"Error: El cupón {cupón_repetido} ya fue utilizado en esta sesión.")
else:
    cupones_usados.add(cupón_repetido)
    print(f"Cupón {cupón_repetido} aplicado con éxito.")
```

Enunciado del Ejercicio:

Deberá construir un programa en Python que resuelva dos casos de uso de la vida real aplicando al menos dos patrones estructurados diferentes aprendidos en la explicación del cuaderno (elija entre LIFO, FIFO o Álgebra de Sets).

Caso 1 (LIFO o FIFO): Diseñe una solución basada en el patrón seleccionado (ejemplo: sistema deshacer/rehacer, atención a clientes en un banco, cola de reproducción musical o gestión de tareas pendientes).


```python
# Simulación de historial de reproduccion musica Utilizando LIFO
historial = []

# Visitar Canciones (Push)
historial.append("No me rendire")
historial.append("animals")
historial.append("perfect")
print("Historial de canciones reproducidas actual:", historial)

# Botón atrás (Pop)
cancion_actual = historial.pop()
print("Saliendo de:", cancion_actual)
print("Cancion donde queda el usuario:", historial[-1])
```

Caso 2 (Gestión con Sets): Implemente un sistema de validación de elementos (ejemplo: control de accesos por roles, filtro de etiquetas duplicadas en publicaciones o verificación de requisitos académicos).

```python
# Lista original de etiquetas ingresadas por el usuario (con duplicados)
etiquetas_base = ["python", "programacion", "python", "codigo", "programacion", "sets"]

# Filtrar duplicados convirtiendo la lista a un conjunto
etiquetas_filtrada = set(etiquetas_base)
print("Etiquetas únicas listas para la publicación:", etiquetas_filtrada)

# Contar cuántas etiquetas duplicadas se eliminaron
cantidad_eliminadas = len(etiquetas_base) - len(etiquetas_filtrada)
print("Cantidad de etiquetas duplicadas removidas:", cantidad_eliminadas)
```
